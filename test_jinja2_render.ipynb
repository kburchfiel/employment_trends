Testing out Jinja2 as a means of rendering an HTML page with both blog text and interactive HTML-based charts:

(The following code was based on https://plotly.com/python/interactive-html-export/#inserting-plotly-output-into-html-using-a-jinja2-template and https://www.geeksforgeeks.org/python/getting-started-with-jinja-template/. Note that the Plotly example won't work in my case due to the %include% statements that I'm using.)

```python
from jinja2 import Environment, FileSystemLoader, select_autoescape
env = Environment(loader=FileSystemLoader(''))
# Here, I'm loading the current directory into my environment,
# as that's where my template and charts folder are located.
```

```python
html_template = env.get_template('test_html_template.html')
```

Rendering the HTML template:

(This code is based on https://www.geeksforgeeks.org/python/getting-started-with-jinja-template/)

```python
rendered_template = html_template.render()
with open('test_rendered_file.html', 'w') as file:
    file.write(rendered_template)
```
